```python
from pyspark.sql.functions import lit
```

```python
%run "../includes/configuration"
```

```python
dbutils.widgets.text("file_date", "2024-12-16", "File Date")
file_date = dbutils.widgets.get("file_date")
```

```python
person_df = (
    spark.read
    .format("json")
    .load(f"{source_folder_path}/{file_date}/person.json")
    .withColumn("file_date", lit(file_date))
    )
```

```python
person_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("replaceWhere", f"file_date = '{file_date}'") \
    .option("path", f"{bronze_folder_path}/external/persons/") \
    .saveAsTable("databricks_ws.bronze.persons")
```
