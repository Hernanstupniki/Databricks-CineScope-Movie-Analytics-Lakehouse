```python
from pyspark.sql.functions import lit
```

```python
%run "../includes/configuration"
```

```python
dbutils.widgets.text("file_date", "2024-12-16", "File Date")
file_date = dbutils.widgets.get("file_date")
```

```python
country_df = (
    spark.read
    .format("json")
    .load(f"{source_folder_path}/{file_date}/country.json")
    .withColumn("file_date", lit(file_date))
    )
```

```python
country_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option(
        "path",
        f"{bronze_folder_path}/external/countries/"
    ) \
    .saveAsTable("databricks_ws.bronze.countries")
```
