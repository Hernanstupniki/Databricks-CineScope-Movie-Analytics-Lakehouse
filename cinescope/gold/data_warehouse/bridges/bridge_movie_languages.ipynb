```sql
USE databricks_ws.silver;
```

```sql
CREATE TABLE IF NOT EXISTS databricks_ws.gold.bridge_movie_languages
USING DELTA
AS
SELECT
    movie_id,
    language_id
FROM databricks_ws.silver.movie_languages
WHERE 1 = 0;
```

```sql
MERGE INTO databricks_ws.gold.bridge_movie_languages AS tgt
USING (
    SELECT
        ml.movie_id,
        ml.language_id
    FROM databricks_ws.silver.movie_languages ml
    INNER JOIN databricks_ws.gold.dim_movies m
        ON ml.movie_id = m.movie_id
    WHERE ml.file_date = :file_date
) AS src
ON tgt.movie_id = src.movie_id AND tgt.language_id = src.language_id

WHEN MATCHED THEN
    UPDATE SET *
WHEN NOT MATCHED THEN
    INSERT *;
```

```sql
SELECT * FROM databricks_ws.gold.bridge_movie_languages;
```
