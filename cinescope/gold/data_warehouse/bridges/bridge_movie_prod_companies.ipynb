```sql
USE databricks_ws.silver;
```

```sql
CREATE TABLE IF NOT EXISTS databricks_ws.gold.bridge_movie_prod_companies
USING DELTA
AS
SELECT
    movie_id,
    company_id
FROM databricks_ws.silver.movie_companies
WHERE 1 = 0;
```

```sql
MERGE INTO databricks_ws.gold.bridge_movie_prod_companies AS tgt
USING (
    SELECT
        mc.movie_id,
        mc.company_id
    FROM databricks_ws.silver.movie_companies mc
    INNER JOIN databricks_ws.gold.dim_movies m
        ON mc.movie_id = m.movie_id
    WHERE mc.file_date = :file_date
) AS src
ON tgt.movie_id = src.movie_id AND tgt.company_id = src.company_id

WHEN MATCHED THEN
    UPDATE SET *
WHEN NOT MATCHED THEN
    INSERT *;
```

```sql
SELECT * FROM databricks_ws.gold.bridge_movie_prod_companies;
```
