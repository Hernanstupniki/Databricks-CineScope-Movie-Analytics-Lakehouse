```sql
CREATE OR REPLACE TABLE databricks_ws.gold.dim_time
USING DELTA
AS
SELECT DISTINCT
    year_release_date AS time_id,
    year_release_date AS release_year,
    CAST(FLOOR(year_release_date / 10) * 10 AS INT) AS decade
FROM databricks_ws.silver.movies
WHERE year_release_date IS NOT NULL;
```

```sql
SELECT *
FROM databricks_ws.gold.dim_time
ORDER BY release_year;
```
