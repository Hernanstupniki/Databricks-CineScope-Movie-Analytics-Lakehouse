```sql
CREATE TABLE IF NOT EXISTS databricks_ws.gold.dim_persons
USING DELTA
AS
SELECT
    person_id,
    name
FROM databricks_ws.silver.persons
WHERE 1 = 0;
```

```sql
MERGE INTO databricks_ws.gold.dim_persons AS tgt
USING (
    SELECT
        person_id,
        name
    FROM databricks_ws.silver.persons
    WHERE file_date = :file_date
) AS src
ON tgt.person_id = src.person_id

WHEN MATCHED THEN
    UPDATE SET *
WHEN NOT MATCHED THEN
    INSERT *;
```
