```sql
USE databricks_ws.silver;
```

```sql
CREATE TABLE IF NOT EXISTS databricks_ws.gold.dim_movies
USING DELTA
AS
SELECT
    movie_id,
    title,
    release_date,
    year_release_date,
    duration_time
FROM databricks_ws.silver.movies
WHERE 1 = 0;
```

```sql
MERGE INTO databricks_ws.gold.dim_movies AS tgt
USING (
    SELECT
        movie_id,
        title,
        release_date,
        year_release_date,
        duration_time
    FROM databricks_ws.silver.movies
    WHERE file_date = :file_date
) AS src
ON tgt.movie_id = src.movie_id

WHEN MATCHED THEN
    UPDATE SET *
WHEN NOT MATCHED THEN
    INSERT *;
```

```sql
USE databricks_ws.gold;
SELECT * FROM dim_movies;
```
