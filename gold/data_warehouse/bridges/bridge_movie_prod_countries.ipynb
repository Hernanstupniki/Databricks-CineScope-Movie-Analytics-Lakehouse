```sql
USE databricks_ws.silver;
```

```sql
CREATE TABLE IF NOT EXISTS databricks_ws.gold.bridge_movie_prod_countries
USING DELTA
AS
SELECT
    movie_id,
    country_id
FROM databricks_ws.silver.production_countries
WHERE 1 = 0;
```

```sql
MERGE INTO databricks_ws.gold.bridge_movie_prod_countries AS tgt
USING (
    SELECT
        pc.movie_id,
        pc.country_id
    FROM databricks_ws.silver.production_countries pc
    INNER JOIN databricks_ws.gold.dim_movies m
        ON pc.movie_id = m.movie_id
    WHERE pc.file_date = :file_date
) AS src
ON tgt.movie_id = src.movie_id AND tgt.country_id = src.country_id

WHEN MATCHED THEN
    UPDATE SET *
WHEN NOT MATCHED THEN
    INSERT *;
```

```sql
SELECT * FROM databricks_ws.gold.bridge_movie_prod_countries;
```
