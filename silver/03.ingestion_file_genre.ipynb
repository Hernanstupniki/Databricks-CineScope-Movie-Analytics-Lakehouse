```python
from pyspark.sql.functions import col, to_date, current_timestamp, lit
```

```python
genres_df = spark.table("databricks_ws.bronze.genres")
```

```python
genres_selected_df = (
    genres_df
    .select(
        col("genreId").cast("int"),
        col("genreName").cast("string")
    )
)
```

```python
genres_final_df = (
    add_ingestion_date(genres_selected_df)
    .withColumnRenamed("genreId", "genre_id")
    .withColumnRenamed("genreName", "genre_name")
    .withColumn("environment", lit("Production"))
)
```

```python
genres_final_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("databricks_ws.silver.genres")
```
