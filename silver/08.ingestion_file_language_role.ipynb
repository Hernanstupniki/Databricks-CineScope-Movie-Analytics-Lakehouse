```python
from pyspark.sql.functions import col, current_timestamp, lit
```

```python
language_role_df = spark.table("databricks_ws.bronze.language_roles")
```

```python
language_role_selected_df = (
    language_role_df.select(
        col("roleId").cast("integer"),
        col("languageRole").cast("string")
        )
    )
```

```python
language_role_final_df = (
    add_ingestion_date(language_role_selected_df)
    .withColumnRenamed("roleId", "role_id")
    .withColumnRenamed("languageRole", "language_role")
    .withColumn("environment", lit("Production"))
)
```

```python
language_role_final_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("databricks_ws.silver.language_roles")
```
