```python
from pyspark.sql.functions import col, to_date, current_timestamp, lit
```

```python
languagues_df = spark.table("databricks_ws.bronze.languages")
```

```python
languages_selected_df = (
    languagues_df
    .select(
        col("languageId").cast("int"),
        col("languageCode").cast("string"),
        col("languageName").cast("string")
    )
)
```

```python
languages_final_df = (
    add_ingestion_date(languages_selected_df)
    .withColumnRenamed("languageId", "language_id")
    .withColumnRenamed("languageCode", "language_code")
    .withColumnRenamed("languageName", "language_name")
    .withColumn("environment", lit("Production"))
)
```

```python
languages_final_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("databricks_ws.silver.languages")
```
