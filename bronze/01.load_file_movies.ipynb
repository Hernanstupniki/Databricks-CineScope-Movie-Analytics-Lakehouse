```python
from pyspark.sql.functions import lit
```

```python
%run "../includes/configuration"
```

```python
dbutils.widgets.text("file_date", "2024-12-16", "File Date")
file_date = dbutils.widgets.get("file_date")
```

```python
movie_df = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("multiLine", "true")
    .option("quote", '"')
    .option("escape", '"')
    .load(f"{source_folder_path}/{file_date}/movie.csv")
    .withColumn("file_date", lit(file_date))
)
```

```python
movie_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("replaceWhere", f"file_date = '{file_date}'") \
    .option("path", f"{bronze_folder_path}/external/movies/") \
    .saveAsTable("databricks_ws.bronze.movies")
```
