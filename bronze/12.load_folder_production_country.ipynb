```python
from pyspark.sql.functions import lit
```

```python
%run "../includes/configuration"
```

```python
dbutils.widgets.text("file_date", "2024-12-16", "File Date")
file_date = dbutils.widgets.get("file_date")
```

```python
production_country_df = (
    spark.read
    .format("json")
    .option("multiLine", True)
    .load(f"{source_folder_path}/{file_date}/production_country")
    .withColumn("file_date", lit(file_date)
    )
)
```

```python
production_country_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("replaceWhere", f"file_date = '{file_date}'") \
    .option("path", f"{bronze_folder_path}/external/production_countries/") \
    .saveAsTable("databricks_ws.bronze.production_countries")
```
